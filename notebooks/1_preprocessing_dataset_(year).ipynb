# Preprocessing of the whole dataset

Stop-word removal and lemmatization are applied to articles from the entire dataset for a one-year period (no filtering by sections to maximize training data).

The vocabulary size is reduced because we are working with a small quantity of documents.  Before training the word2vev model, we will remove stop-words and lemmatize the text.

- input: dataset (TSV) (not filtered by category)
- output: preprocessed dataset

Library imports

```python
import pandas as pd
import spacy

from tqdm import tqdm
tqdm.pandas() # This is to show progress bars in google colab
```

Load language model

```python
nlp = spacy.load('ru_core_news_md', disable=["ner", "parser"])  # Disable unused components for speed
```

Read dataset

```python
year = 2022

df = pd.read_csv(f'pravda_{year}.tsv', sep='\t')
df
```

A function to preprocess the documents

```python
def lemmatize(document):
  lemmas = list()
  doc = nlp(document)
  for token in doc:
    if token.is_punct == False and token.is_stop == False:
      lemma = token.lemma_
      lemmas.append(lemma)
  return ' '.join(lemmas)
```

Preprocessing

```python
df['lemmatized_paragraphs'] = df['paragraphs'].progress_apply(lemmatize)
```

Save the lemmatized dataset

```python
output_file = f'pravda_lemmatized_{year}.tsv'

df.to_csv(output_file, sep='\t', index=False)
```
