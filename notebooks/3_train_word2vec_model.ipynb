- Output: a file with the extension ( .model)

Library imports

```python
from gensim.models import Word2Vec
import json
```

Load the tokenized documents

```python
year = 2022

file_path = f'/content/sample_data/tokenized_documents_{year}.json'

with open(file_path, 'r') as f:
  tokenized_documents = json.load(f)
```

Train the Word2Vec model

```python
model = Word2Vec(
    sentences=tokenized_documents,  # Tokenized input
    vector_size=100,                # Size of word vectors (e.g., 100 dimensions)
    window=5,                       # Context window size
    min_count=1,                    # Ignore words with frequency < 1
    sg=0,                           # 0 for CBOW, 1 for Skip-gram
    epochs=10,                      # Number of training iterations
    workers=4                       # Number of CPU cores to use
)
```

A function to zip the model and its associated files

```python
def save_model_as_zip(model, folder_name, output_file):
    if not os.path.exists(folder_name):
        os.makedirs(folder_name)

    model.save(os.path.join(folder_name, f'{output_file}.model'))

    shutil.make_archive(output_file, 'zip', root_dir=folder_name)

    # 4. Delete the folder after zipping
    shutil.rmtree(folder_name)
```

Save the trained Word2vec model

```python
output_file = f"word2vec_{year}"

save_model_as_zip(model, f'model_{year}', output_file)
```
