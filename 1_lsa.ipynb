# Latent Semantic Analysis (LSA)

Latent Semantic Analysis (LSA) is useful for documents because it uncovers hidden semantic relationships by reducing high-dimensional term-document matrices into a lower-dimensional space, capturing latent topics and improving text analysis. By applying singular value decomposition (SVD), LSA identifies patterns in word co-occurrences, allowing it to group documents with similar meanings

Requirements: preprocessed dataframe

Import libraries

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
```

Load the preprocessed dataset

```python
year = 2022
category = 'economics'

df = pd.read_csv(f'/content/sample_data/pravda_preprocessed_{year}_{category}.tsv', sep='\t')
df
```

```python
documents = df['lemmatized_paragraphs'].tolist()
```

```python
documents[0]
```

Create Document-Term Matrix

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)
tfidf_matrix
```

Apply LSA

```python
n_topics = 10 # Experiment with different numbers

svd = TruncatedSVD(n_components=n_topics, random_state=42)
lsa_matrix = svd.fit_transform(tfidf_matrix)
```

Show the most significant terms for each topic

```python
top_n = 10

terms = vectorizer.get_feature_names_out()
for i, component in enumerate(svd.components_):
    top_terms = [terms[idx] for idx in component.argsort()[-top_n:][::-1]]
    topic = f"Topic {i+1}: " + ", ".join(top_terms)
    print(topic)
```
